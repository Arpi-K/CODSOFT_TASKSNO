```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data=pd.read_csv('HR_Analytics-4.csv')
data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
data.isnull().sum()
```

```python
data.dropna(inplace=True)
```

```python
data.columns
```

```python
data['BusinessTravel'].unique()
```

```python
data['BusinessTravel']=data['BusinessTravel'].str.replace(
    {'TravelRarely':'Travel_Rarely',
     'Travel-Frequently':'Travel_Frequently'})
```

## DATA VISUALIZATION

```python
sns.set_theme(style='whitegrid')
```

```python
plt.figure(figsize=(7,5))
sns.countplot(data=data,x='Attrition')
plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=data, x="Gender", hue="Attrition")
plt.title("Employee Attrition by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Employees")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.countplot(data=data, x="Department", hue="Attrition")
plt.title("Employee Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=15)
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.histplot(data=data, x="MonthlyIncome", bins=30, kde=True)
plt.title("Distribution of Monthly Income")
plt.xlabel("Monthly Income")
plt.ylabel("Number of Employees")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.histplot(data=data, x="Age", bins=20, kde=True)
plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```

```python
plt.figure(figsize=(9,6))
sns.scatterplot(
    data=data,
    x="Age",
    y="MonthlyIncome",
    hue="Attrition"
)
plt.title("Age vs Monthly Income by Attrition")
plt.xlabel("Age")
plt.ylabel("Monthly Income")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=data, x="JobSatisfaction")
plt.title("Employee Job Satisfaction")
plt.xlabel("Job Satisfaction Level")
plt.ylabel("Number of Employees")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.countplot(
    data=data,
    x="JobSatisfaction",
    hue="Attrition"
)
plt.title("Job Satisfaction vs Employee Attrition")
plt.xlabel("Job Satisfaction Level")
plt.ylabel("Number of Employees")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(
    data=data,
    x="OverTime",
    hue="Attrition"
)
plt.title("Overtime vs Employee Attrition")
plt.xlabel("Overtime")
plt.ylabel("Number of Employees")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.countplot(
    data=data,
    y="JobRole",
    hue="Attrition"
)
plt.title("Employee Attrition by Job Role")
plt.xlabel("Number of Employees")
plt.ylabel("Job Role")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=data,
    x="BusinessTravel",
    hue="Attrition"
)

plt.title("Business Travel vs Employee Attrition")
plt.xlabel("Business Travel")
plt.ylabel("Number of Employees")

plt.show()
```

```python
numeric_df = data.select_dtypes(include=np.number)
plt.figure(figsize=(16,12))
sns.heatmap(
    numeric_df.corr(),
    annot=False,
    cmap="coolwarm"
)
plt.title("Correlation Heatmap of HR Analytics Dataset")
plt.show()
```
